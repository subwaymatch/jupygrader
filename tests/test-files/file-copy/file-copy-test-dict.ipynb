## Case 1: Copy to the same directory

```python
with open('copied1.txt', 'r') as f:
    s1 = f.read()

print(s1)
```

```python
import unittest

tc = unittest.TestCase()

_test_case = 'tc-read-files-01'
_points = 10

tc.assertEqual(s1, 'This is a text file')
```
